```python
import pandas as pd
from sqlalchemy import create_engine, text
from urllib.parse import quote_plus

# Step 3: Enter your TiDB connection details
host = "gateway01.ap-northeast-1.prod.aws.tidbcloud.com"
port = 4000
user = "3k5gspKEptkYxvs.root"
password = "uJVYvVceTGGGT00O"
database = "tmdb_proj_vilasini"

# Step 4: Create the database connection
connection_url = (f"mysql+pymysql://{user}:{password}"
    f"@{host}:{port}/{database}"
    "?ssl_verify_cert=true&ssl_verify_identity=true"
)
engine = create_engine(connection_url, pool_recycle=300)

#SQL 9: Find the genre with the highest average rating, considering only genres with at least 20 movies.	JOIN + GROUP BY + HAVING + AVG()
with engine.connect() as conn:
    result = conn.execute(
        text(""" select a.genre_id, a.genre_name, count(distinct b.movie_id) as movie_count, avg(b.vote_average) as avg_rating
from cleaned_generes a,
cleaned_movies b,
cleaned_movie_generes c
where a.genre_id = c.genre_id
and c.movie_id = b.movie_id
group by a.genre_id,a.genre_name
having count(distinct b.movie_id) > 20
order by avg_rating desc"""))

 # Convert SQL result into Pandas DataFrame
    df_result = pd.DataFrame(
        result.fetchall(),
        columns=result.keys()
    )

# Display interactive database-style table
display(df_result)

print("Total Records:", len(df_result))

engine.dispose()
```
