```python
#1. List the total number of unique movies, genres, cast members and crew members collected across all tables.
import pandas as pd

# Step 1: Load the CSV files
movies_df = pd.read_csv("movies.csv")
genres_df = pd.read_csv("genres.csv")
cast_df = pd.read_csv("cast.csv")
crew_df = pd.read_csv("crew.csv")

# Step 2: Count unique records
total_movies = movies_df["movie_id"].nunique()
total_genres = genres_df["genre_id"].nunique()
total_cast_members = cast_df["person_id"].nunique()
total_crew_members = crew_df["person_id"].nunique()

# Step 3: Display the results
print("Total unique movies:", total_movies)
print("Total unique genres:", total_genres)
print("Total unique cast members:", total_cast_members)
print("Total unique crew members:", total_crew_members)
```

```python
#2. (i) List all movies with missing or zero budget/revenue values,


import pandas as pd

# Load the movies dataset
df = pd.read_csv("movies.csv")

# Find movies with missing or zero budget/revenue
condition = (df["budget"].isna() |  (df["budget"] == 0) | df["revenue"].isna() | (df["revenue"] == 0))

missing_movies = df.loc[condition,["movie_id", "budget", "revenue"]]

# Display all matching movies
print(missing_movies.to_string(index=False))

# Count affected movies
print("\nTotal movies with missing or zero budget/revenue:", len(missing_movies))

# (ii) and decide how these should be treated for the rest of the analysis.

budget_mean = df.loc[(df["budget"].notna()) & (df["budget"] != 0), "budget"].mean()
revenue_mean = df.loc[(df["revenue"].notna()) & (df["revenue"] != 0),"revenue"].mean()

df["budget"] = df["budget"].fillna(budget_mean)
df["revenue"] = df["revenue"].fillna(revenue_mean)

# Step 4: Verify the result
print("Remaining null budgets:", df["budget"].isna().sum())
print("Remaining null revenues:", df["revenue"].isna().sum())

print("Remaining zero budgets:", (df["budget"] == 0).sum())
print("Remaining zero revenues:", (df["revenue"] == 0).sum())
```

```python
#3 List the profit (revenue − budget) and return on investment for every movie that has valid budget and revenue figures.
import pandas as pd

df=pd.read_csv("movies.csv")

# Keep movies with valid, positive budget and revenue
valid_movies = df[
    df["budget"].notna() &
    df["revenue"].notna() &
    (df["budget"] > 0) &
    (df["revenue"] > 0)
].copy()

# Calculate profit
valid_movies["profit"] = valid_movies["revenue"] - valid_movies["budget"]

# Calculate ROI as a percentage
valid_movies["ROI"] = (
    (valid_movies["revenue"] - valid_movies["budget"])
    / valid_movies["budget"]
) * 100

# Display movie title, budget, revenue, profit, and ROI
print(valid_movies[
    ["title", "budget", "revenue", "profit", "ROI"]
].head(10))
```

```python
#4 List the top 15 movies by return on investment, considering only movies with a budget of at least $1 million.
import pandas as pd

df=pd.read_csv("movies.csv")

# Filter movies with valid budget and revenue
valid_movies = df[
    df["budget"].notna() &
    df["revenue"].notna() &
    (df["budget"] >= 1_000_000) &
    (df["revenue"] > 0)
].copy()

# Calculate ROI as a percentage
valid_movies["ROI"] = (
    (valid_movies["revenue"] - valid_movies["budget"])
    / valid_movies["budget"]
) * 100

# Sort by ROI in descending order and select top 15
top_15_roi = valid_movies.sort_values(
    by="ROI", ascending=False
).head(15)

# Display results
print(top_15_roi[["title", "budget", "revenue", "ROI"]])
```

```python
#5 List the number of movies released per year, and identify which year had the highest movie output in the dataset.
import pandas as pd

df=pd.read_csv("movies.csv")

# Convert release_date to datetime
df["release_date"] = pd.to_datetime(df["release_date"], errors="coerce")

# Extract release year
df["release_year"] = df["release_date"].dt.year

# Count movies released per year
movies_per_year = df["release_year"].value_counts().sort_index()

print("Number of movies released per year:")
print(movies_per_year)

# Identify the year with the highest movie output
highest_year = movies_per_year.idxmax()
highest_count = movies_per_year.max()

print("\nYear with the highest movie output:", highest_year)
print("Number of movies released:", highest_count)
```

```python
#6 List all movies whose runtime is a statistical outlier compared to the rest of the dataset.
import pandas as pd

df=pd.read_csv("movies.csv")

# Keep movies with valid runtime values
runtime_df = df[df["runtime"].notna() & (df["runtime"] > 0)].copy()

# Calculate Q1, Q3, and IQR
Q1 = runtime_df["runtime"].quantile(0.25)
Q3 = runtime_df["runtime"].quantile(0.75)
IQR = Q3 - Q1

# Calculate outlier boundaries
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify runtime outliers
outliers = runtime_df[
    (runtime_df["runtime"] < lower_bound) |
    (runtime_df["runtime"] > upper_bound)
]

# Display outlier movies
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
print("\nMovies with outlier runtimes:")
print(outliers[["title", "runtime"]].sort_values(
    by="runtime", ascending=False
))
```

```python
#7 List each movie's primary genre (the first genre listed) and compare average popularity across primary genres.
import pandas as pd

# Read the CSV files
movies = pd.read_csv("movies.csv")
movie_genres = pd.read_csv("movie_genres.csv")
genres = pd.read_csv("genres.csv")

# Merge movies with the movie-genre mapping
df = movies.merge(movie_genres, on="movie_id", how="left")

# Merge with the genre details
df = df.merge(genres, on="genre_id", how="left")

# Select the first genre listed for each movie
# Assumes movie_genres.csv preserves the intended genre order
df["primary_genre"] = df.groupby("movie_id")["genre_name"].transform("first")

# List each movie with its primary genre and popularity
movie_genre_list = df[
    ["movie_id", "title", "primary_genre", "popularity"]
].drop_duplicates(subset="movie_id")

print("Each movie and its primary genre:")
print(movie_genre_list)

# Compare average popularity across primary genres
avg_popularity = (
    movie_genre_list
    .dropna(subset=["primary_genre", "popularity"])
    .groupby("primary_genre")["popularity"]
    .mean()
    .sort_values(ascending=False)
)

print("\nAverage popularity by primary genre:")
print(avg_popularity)
```

```python
#8 List the top 10 most prolific actors by number of movies, along with their average movie rating across those movies.
import pandas as pd

# Read CSV files
movies = pd.read_csv("movies.csv")
cast = pd.read_csv("cast.csv")

# Merge cast data with movie ratings
df = cast.merge(
    movies[["movie_id", "title", "vote_average"]],
    on="movie_id",
    how="inner"
)

# Remove duplicate actor-movie combinations
df = df.drop_duplicates(subset=[ "movie_id"])

# Group by actor and calculate movie count and average rating
actor_stats = (
    df.groupby(["actor_name"])
    .agg(
        number_of_movies=("movie_id", "nunique"),
        average_rating=("vote_average", "mean")
    )
    .reset_index()
)

# Sort by number of movies and select top 10
top_10_actors = actor_stats.sort_values(
    by=["number_of_movies", "average_rating"],
    ascending=[False, False]
).head(10)

# Display results
print(top_10_actors.to_string(index=False))
```

```python
#9 List any movie titles that appear more than once in the dataset, along with their release years, to check for duplicates or remakes.
import pandas as pd
from google.colab import data_table

# Enable interactive table output
data_table.enable_dataframe_formatter()

# Read CSV file
movies = pd.read_csv("movies.csv")

# Extract release year
movies["release_year"] = pd.to_datetime(
    movies["release_date"], errors="coerce"
).dt.year

# Find movie titles appearing more than once
duplicate_movies = movies[
    movies.duplicated(subset=["title"], keep=False)
]

# Group by title and release year
result = duplicate_movies.groupby(
    ["title", "release_year"]
).agg(
    total_count=("movie_id", "count")
).reset_index()

# Sort the results
result = result.sort_values(
    by=["title", "release_year"]
)

# Display interactive table
display(result)

print("Total duplicate title-year records:", len(result))
```

```python
#10 - List movies whose popularity score and vote count don't follow the general pattern between the two — i.e., unusually high popularity with a low vote count, or vice versa.
import pandas as pd
from google.colab import data_table

# Enable interactive table output
data_table.enable_dataframe_formatter()

# Read CSV file
movies = pd.read_csv("movies.csv")

# Select required columns and remove missing values
df = movies[[
    "movie_id", "title", "popularity", "vote_count"
]].dropna()

# Remove invalid values
df = df[
    (df["popularity"] >= 0) &
    (df["vote_count"] >= 0)
].copy()

# Calculate percentile thresholds
high_popularity = df["popularity"].quantile(0.90)
low_popularity = df["popularity"].quantile(0.10)

high_votes = df["vote_count"].quantile(0.90)
low_votes = df["vote_count"].quantile(0.10)

# Identify unusual combinations
df["observation"] = "Normal"

df.loc[
    (df["popularity"] >= high_popularity) &
    (df["vote_count"] <= low_votes),
    "observation"
] = "High Popularity - Low Votes"

df.loc[
    (df["popularity"] <= low_popularity) &
    (df["vote_count"] >= high_votes),
    "observation"
] = "Low Popularity - High Votes"

# Filter unusual movies
result = df[df["observation"] != "Normal"]

# Sort by popularity
result = result.sort_values(
    by="popularity",
    ascending=False
)

# Display interactive table
display(result[[
    "movie_id",
    "title",
    "popularity",
    "vote_count",
    "observation"
]])

print("Total unusual movies:", len(result))
```
